```python
import pandas as pd

df_ecom = pd.read_csv('OnlineRetail.csv', encoding='ISO-8859-1')
df_ecom.head()
```

```python
df_netflix = pd.read_csv('netflix_titles.csv')
df_netflix.head()
```

```python
print("Shape (rows, columns):", df_ecom.shape)
print("\nColumn data types:")
print(df_ecom.dtypes)
print("\nMissing values per column:")
print(df_ecom.isnull().sum())
```

```python
print("Shape (rows, columns):", df_netflix.shape)
print("\nColumn data types:")
print(df_netflix.dtypes)
print("\nMissing values per column:")
print(df_netflix.isnull().sum())
```

```python
print("Duplicate rows in E-commerce dataset:", df_ecom.duplicated().sum())
print("Duplicate rows in Netflix dataset:", df_netflix.duplicated().sum())
```

```python
before = df_ecom.shape[0]
df_ecom = df_ecom.drop_duplicates()
after = df_ecom.shape[0]

print("Rows before:", before)
print("Rows after removing duplicates:", after)
print("Duplicates removed:", before - after)
```

```python
# Drop rows where Description is missing
df_ecom = df_ecom.dropna(subset=['Description'])

# Confirm missing values now
print(df_ecom.isnull().sum())
```

```python

df_ecom['InvoiceDate'] = pd.to_datetime(df_ecom['InvoiceDate'])
print(df_ecom.dtypes)
```

```python
print("Rows with negative Quantity:", (df_ecom['Quantity'] < 0).sum())
print("Rows with negative or zero UnitPrice:", (df_ecom['UnitPrice'] <= 0).sum())
```

```python
# Remove rows with zero or negative price (invalid entries)
before = df_ecom.shape[0]
df_ecom = df_ecom[df_ecom['UnitPrice'] > 0]
after = df_ecom.shape[0]
print("Rows removed due to invalid UnitPrice:", before - after)

# Separate returns (negative quantity) into their own dataframe, keep only actual sales for main analysis
df_returns = df_ecom[df_ecom['Quantity'] < 0]
df_ecom = df_ecom[df_ecom['Quantity'] > 0]

print("Returns separated:", df_returns.shape[0])
print("Remaining sales rows:", df_ecom.shape[0])
```

```python
# Standardize text: remove extra spaces, make column names lowercase for consistency
df_ecom.columns = df_ecom.columns.str.strip().str.lower()
df_ecom['description'] = df_ecom['description'].str.strip().str.upper()
df_ecom['country'] = df_ecom['country'].str.strip()

df_ecom.head()
```

```python
# Fill missing categorical info with "Unknown"
df_netflix['director'] = df_netflix['director'].fillna('Unknown')
df_netflix['cast'] = df_netflix['cast'].fillna('Unknown')
df_netflix['country'] = df_netflix['country'].fillna('Unknown')

# Drop rows with missing date_added, rating, or duration (small number, safe to drop)
before = df_netflix.shape[0]
df_netflix = df_netflix.dropna(subset=['date_added', 'rating', 'duration'])
after = df_netflix.shape[0]

print("Rows dropped due to missing date_added/rating/duration:", before - after)
print(df_netflix.isnull().sum())
```

```python
# Convert date_added to proper datetime
df_netflix['date_added'] = pd.to_datetime(df_netflix['date_added'].str.strip())

# Standardize column names and text
df_netflix.columns = df_netflix.columns.str.strip().str.lower()
df_netflix['type'] = df_netflix['type'].str.strip()
df_netflix['country'] = df_netflix['country'].str.strip()

df_netflix.dtypes
```

## Key Insights — E-commerce Dataset

1. **A few products are bought in bulk**: Items like "Paper Craft, Little Birdie" and "Medium Ceramic Top Storage Jar" sold the most in total quantity. But this is probably because of one or two big wholesale orders, not because a lot of different customers bought them (Chart 1).

2. **Almost all the money comes from the UK**: The United Kingdom brought in over £9 million, way more than every other country put together. The Netherlands, the next best, made less than £300,000. This shows the business relies heavily on UK customers (Chart 2).

3. **Sales go up towards the end of the year**: Revenue kept growing through 2011 and was highest in November, most likely because of holiday shopping. This means the business should prepare more stock and marketing around that period ( Chart 3).

4. **Some products are popular with many people, not just bulk buyers**: "White Hanging Heart T-Light Holder" was ordered the most times across different orders, even though it wasn't the top product by quantity. This shows it's genuinely liked by many customers, not just bought in bulk by one or two people (Chart 4).

5. **A small group of customers spend a lot more than everyone else**: Most customers spend less than £2,000 in total, but a small number spend a lot more, some over £10,000. The business could create a loyalty program to keep these big spenders coming back (Chart 5).

## Key Insights — Netflix Dataset

1. **Movies make up most of the content**: About 70% of titles on Netflix are Movies, while only 30% are TV Shows. This means the platform still leans more towards movies than series ( Chart 1).

2. **Netflix added a lot more content after 2015**: Before 2015, very few titles were added each year. But from 2016 to 2019, the numbers grew very fast, reaching over 2,000 titles added in 2019 alone. After that, it dropped a bit in 2020 and 2021, but this might just be because the dataset doesn't cover the full year 2021 (Chart 2).

3. **The United States produces the most content**: The US has way more titles than any other country, followed by India and the UK. This shows Netflix relies a lot on US-made content, but is also investing in other countries like India ( Chart 3).

4. **Most content is made for older audiences**: The most common rating is TV-MA (mature audience), followed by TV-14. This means Netflix has more content for adults and teens than for young children ( Chart 4).

5. **International movies and dramas are the most common genres**: "International Movies" and "Dramas" appear the most, followed by "Comedies" and "International TV Shows." This shows Netflix focuses a lot on international and dramatic stories (Chart 5).
